```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.document_loaders import PyPDFLoader

# Load the PDF
pdf_path = "C:\\Users\\Gamer Tech\\Desktop\\Medical Info.pdf"
loader = PyPDFLoader(pdf_path)
documents = loader.load()

# Split text into chunks
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
docs = text_splitter.split_documents(documents)

# Save processed chunks (optional)
for i, doc in enumerate(docs[:3]):  # Print first 3 chunks
    print(f"Chunk {i+1}:\n{doc.page_content}\n")
```

```python
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import FAISS
from sentence_transformers import SentenceTransformer


# Convert text chunks into embeddings
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2",)  # You can choose a different model


vector_store = FAISS.from_documents(docs, embeddings)

# Save the FAISS vector store for future use
vector_store.save_local("medical_vector_store")
```

```python
from langchain_ollama import OllamaLLM
from langchain.chains import RetrievalQA
# Load the FAISS vector store
vector_store = FAISS.load_local("medical_vector_store", embeddings, allow_dangerous_deserialization=True)
# Optionally, create a retriever from the vector store
retriever = vector_store.as_retriever()

```

```python
ollama = OllamaLLM(base_url='http://localhost:11434', model="llama3", temperature=0.7)
response=ollama.invoke("am i talking to a llama 2 model or llama 3 model right now? answer in one line")
print(response)
```

```python
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
prompt=PromptTemplate.from_template("what is the capital of {country}?")
qa_chain = RetrievalQA.from_chain_type(ollama, retriever=retriever)

# Ask a medical question
query = "What is the description of diabetes?"
response = qa_chain.run(query)
print(response)
```
